```python
import pandas as pd
```

```python
df=pd.read_csv('/content/psychometric.csv')
```

```python
df.head()
```

```python
X=df[['O','C','E','A','N']]
X.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=100,
    contamination=0.1,
    random_state=42
)

model.fit(X_scaled)
```

```python
df['risk_flag'] = model.predict(X_scaled)
df.head()
```

```python
risky_employees = df[df['risk_flag'] == -1]
risky_employees
```
